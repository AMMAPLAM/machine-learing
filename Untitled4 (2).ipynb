```python
import pandas as pd

data = {
    "Name": ["Ali", "Rahul", "John", "Sara", "Priya"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 72, 90, 65, 88]
}

df = pd.DataFrame(data)

result = df[(df["Marks"] > 80) & (df["Age"] < 21)]

print("Filtered Data:")
print(result)
```

```python
import pandas as pd

data = {
    "Product": ["Laptop", "Mobile", "Tablet", "Headphone", "Monitor"],
    "Price": [60000, 25000, 30000, 5000, 15000],
    "Rating": [4.5, 4.2, 4.0, 4.6, 4.3]
}

df = pd.DataFrame(data)

# Ascending order
print("Ascending Price:")
print(df.sort_values("Price", ascending=True))

# Descending order
print("\nDescending Price:")
print(df.sort_values("Price", ascending=False))
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Rahul", "John", "Sara", "Priya"],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [40000, 35000, 45000, 40000, 38000]
}

df = pd.DataFrame(data)

result = df.sort_values(
    ["Department", "Salary"],
    ascending=[True, False]
)

print(result)
```

```python
import pandas as pd

data = {
    "Customer_ID": [101, 102, 103, 101, 104, 102],
    "Name": ["Ali", "Rahul", "John", "Ali", "Sara", "Rahul"],
    "City": ["Chennai", "Delhi", "Mumbai", "Chennai", "Hyderabad", "Delhi"]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

result = df.drop_duplicates()

print("\nAfter Removing Duplicates:")
print(result)
```

```python
import pandas as pd

data = {
    "Product": ["Laptop", "Mobile", "Laptop", "Tablet", "Mobile"],
    "Sales": [60000, 25000, 55000, 30000, 28000]
}

df = pd.DataFrame(data)

print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())
```

```python
import pandas as pd

data = {
    "Department": ["IT", "IT", "HR", "HR", "Finance"],
    "Employee": ["Ali", "Rahul", "John", "Sara", "Priya"],
    "Salary": [40000, 45000, 35000, 38000, 50000]
}

df = pd.DataFrame(data)

result = df.groupby("Department")["Salary"].mean()

print("Average Salary by Department:")
print(result)
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
temperature = [24, 26, 29, 32, 35, 37]

plt.plot(months, temperature, marker="o")

plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.title("Monthly Temperature")

plt.show()
```

```python
import matplotlib.pyplot as plt

cities = ["Chennai", "Delhi", "Mumbai", "Hyderabad", "Bangalore"]
population = [11, 32, 21, 10, 13]

plt.bar(cities, population)

plt.xlabel("City")
plt.ylabel("Population (Millions)")
plt.title("City Population")

plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [
    45, 52, 60, 65, 70,
    72, 75, 78, 80, 82,
    85, 88, 90, 92, 95
]

plt.hist(marks, bins=5)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [40, 45, 50, 55, 65, 70, 78, 85]

plt.scatter(study_hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ["Food", "Rent", "Travel", "Shopping", "Other"]
expenses = [30, 40, 10, 12, 8]

plt.pie(
    expenses,
    labels=categories,
    autopct="%1.1f%%"
)

plt.title("Monthly Expenses")

plt.show()
```
